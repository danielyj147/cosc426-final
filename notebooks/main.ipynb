# Data Preprocessing

```python
import os
import sys

sys.path.append("..")
import pandas as pd
import requests

# For color coding
RED = "\033[31m"
GREEN = "\033[92m"
BLUE = "\033[94m"
END = "\033[0m"
```

## Download Dataset & Load

```python
# Downloading Dataset

path = "../data/original.csv"
url = "https://opendata.dc.gov/api/download/v1/items/f92f4556f26b4737a040fb996eaefca3/csv?layers=40"

if os.path.exists(path):
    print(f"File already exists at {path}")
else:
    print("File not found. Downloading...")
    response = requests.get(url, stream=True)
    if response.status_code == 200:
        with open(path, "wb") as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
        print(f"File downloaded successfully to {path}")
    else:
        print(f"Failed to download file. HTTP status code: {response.status_code}")
```

```python
# csv => tsv
tsv_path = "../data/original.tsv"
cols = [
    "OBJECTID",
    "RACE",
    "GENDER",
    "AGE_GROUP",
    "OFFENSE",
    "OFFENSE_SEVERITY_GROUP",
    "SENTENCE_TO_SERVE_MONTHS",
]

if not os.path.exists("../data/original.tsv"):
    df = pd.read_csv(path)

    df = df.dropna(subset=cols)
    df.to_csv("../data/original.tsv", sep="\t", index=False)

df = pd.read_csv(tsv_path, sep="\t")
```

## Saving & Loading Offenses

```python
df["OFFENSE"].drop_duplicates().to_csv(
    "../data/offense.csv", index=False, header=["offense"]
)

offense = pd.read_csv("../data/offense.csv")
```
